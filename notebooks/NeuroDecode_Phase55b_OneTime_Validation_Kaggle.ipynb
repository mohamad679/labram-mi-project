## E — Share a small scientific-evidence ZIP; do not share HDF5 or checkpoint binaries

This archive is for independently recalculating the 2,436 original Validation predictions and checking predeclared training, checkpoint, class metrics and subject provenance. Preserve the model weights privately with Kaggle Save Version.


```python
import zipfile,hashlib,json,numpy as np
report=json.loads((RESULTS/'report.json').read_text())
assert report['status']=='COMPLETED_ONE_TIME_DEVELOPMENT_VALIDATION_NO_TEST'
assert report['validation_evaluations']==1
assert report['validation_used_for_early_stopping'] is False
assert report['test_waveform_inference'] is False
assert report['selected_model']=='eegnet_epoch_standardized'
assert len(report['actual_fit_subjects'])==67 and len(report['actual_early_stop_subjects'])==8
assert sorted(report['actual_early_stop_subjects'])==[9,25,56,61,70,81,83,90]
with np.load(RESULTS/'validation_predictions.npz',allow_pickle=False) as a:
    assert a['y_prob'].shape==(2436,5)
    assert len(np.unique(a['subject_ids']))==14
    assert np.all(np.isfinite(a['y_prob']))
    assert np.allclose(a['y_prob'].sum(axis=1),1.,atol=1e-5)
    assert list(a['class_names'])==['feet','hands','left_hand','rest','right_hand']
with zipfile.ZipFile(ZIP,'x',compression=zipfile.ZIP_DEFLATED,compresslevel=6) as z:
    for name in ('report.json','validation_predictions.npz'):
        path=RESULTS/name
        assert path.is_file(),f'Missing scientific evidence: {name}'
        z.write(path,arcname=name)
with zipfile.ZipFile(ZIP) as z:
    assert z.testzip() is None and sorted(z.namelist())==['report.json','validation_predictions.npz']
print('PHASE5.5b ONE-TIME VALIDATION ZIP READY:',ZIP)
print('ZIP size MiB:',round(ZIP.stat().st_size/2**20,3))
print('ZIP SHA-256:',sha(ZIP))
print('Preserve selected checkpoint in private Kaggle Saved Version; do not upload raw EEG.')
```
